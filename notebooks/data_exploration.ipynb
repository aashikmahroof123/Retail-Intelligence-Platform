```python
top_products = (
    data.groupby("ProductName")["Quantity"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print(top_products)
```

```python
top_revenue_products = (
    data.groupby("ProductName")["TotalAmount"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print(top_revenue_products)
```

```python
category_sales = (
    data.groupby("Category")["TotalAmount"]
    .sum()
    .sort_values(ascending=False)
)

print(category_sales)
```

```python
print("Average stock level:", data["StockQuantity"].mean())
print("Minimum stock level:", data["StockQuantity"].min())
print("Maximum stock level:", data["StockQuantity"].max())
```

```python
low_stock = data[data["StockQuantity"] < 20]

print(low_stock[[
    "ProductID",
    "ProductName",
    "StockQuantity"
]].head(20))
```

```python
data["InvoiceDate"] = pd.to_datetime(data["InvoiceDate"])

print("First transaction:", data["InvoiceDate"].min())
print("Last transaction:", data["InvoiceDate"].max())
```

```python
data["Year"] = data["InvoiceDate"].dt.year
data["Month"] = data["InvoiceDate"].dt.month
data["Day"] = data["InvoiceDate"].dt.day
data["DayOfWeek"] = data["InvoiceDate"].dt.day_name()
```

```python
monthly_sales = (
    data.groupby("Month")["TotalAmount"]
    .sum()
)

print(monthly_sales)
```

```python
daily_sales = (
    data.groupby(data["InvoiceDate"].dt.date)["TotalAmount"]
    .sum()
)

print(daily_sales.head(10))
```

```python
day_sales = (
    data.groupby(data["InvoiceDate"].dt.day_name())["TotalAmount"]
    .sum()
)

print(day_sales)
```

```python
store_sales = (
    data.groupby("StoreName")["TotalAmount"]
    .sum()
    .sort_values(ascending=False)
)

print(store_sales)
```

```python
products_per_invoice = (
    data.groupby("InvoiceID")["ProductID"]
    .nunique()
)

print(products_per_invoice.describe())
```

```python
customer_orders = (
    data.groupby("CustomerID")["InvoiceID"]
    .nunique()
)

print(customer_orders.describe())
```

```python
repeat_customers = (
    customer_orders > 1
).sum()

total_customers = data["CustomerID"].nunique()

print("Repeat customers:", repeat_customers)
print("Total customers:", total_customers)
```

```python
customer_sales = (
    data.groupby("CustomerID")["TotalAmount"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print(customer_sales)
```
