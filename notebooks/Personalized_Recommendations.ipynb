```python
import pandas as pd

customer_product = pd.read_excel(
    "../datasets/customer_product_features.xlsx"
)

product_features = pd.read_excel(
    "../datasets/product_features.xlsx"
)

print("Customer-product columns:")
print(customer_product.columns.tolist())

print()

print("Product feature columns:")
print(product_features.columns.tolist())
```

```python
print(
    customer_product.isnull().sum()
)
```

```python
customer_product = customer_product.dropna(
    subset=[
        "CustomerID",
        "ProductID"
    ]
).copy()

print(
    "Rows after cleaning:",
    len(customer_product)
)
```

```python
customer_product["CustomerID"] = (
    customer_product["CustomerID"]
    .astype(str)
)

customer_product["ProductID"] = (
    customer_product["ProductID"]
    .astype(str)
)
```

```python
print(
    customer_product["TotalQuantity"].describe()
)
```

```python
purchase_matrix = customer_product.pivot_table(
    index="CustomerID",
    columns="ProductID",
    values="TotalQuantity",
    aggfunc="sum",
    fill_value=0
)

purchase_matrix.head()
```

```python
print(
    "Customers:",
    purchase_matrix.shape[0]
)

print(
    "Products:",
    purchase_matrix.shape[1]
)
```

```python
similarity_matrix = cosine_similarity(
    purchase_matrix
)

print(
    "Similarity matrix shape:",
    similarity_matrix.shape
)
```

```python
similarity = pd.DataFrame(
    similarity_matrix,
    index=purchase_matrix.index,
    columns=purchase_matrix.index
)

similarity.head()
```

```python
def recommend_for_customer(
    customer_id,
    limit=5
):

    customer_id = str(customer_id)

    if customer_id not in purchase_matrix.index:
        return []

    similar_customers = (
        similarity[customer_id]
        .drop(index=customer_id)
        .sort_values(
            ascending=False
        )
    )

    scores = (
        purchase_matrix
        .loc[similar_customers.index]
        .T
        .dot(similar_customers)
    )

    bought = purchase_matrix.loc[
        customer_id
    ]

    scores = scores[
        bought == 0
    ]

    scores = scores[
        scores > 0
    ]

    scores = scores.sort_values(
        ascending=False
    ).head(limit)

    return [
        {
            "ProductID": product_id,
            "Score": round(
                float(score),
                4
            )
        }
        for product_id, score
        in scores.items()
    ]
```

```python
test_customer = (
    purchase_matrix.index[0]
)

print(
    "Testing customer:",
    test_customer
)
recommendations = recommend_for_customer(
    test_customer,
    limit=5
)

recommendations
```

```python
test_customers = (
    purchase_matrix.index[:10]
)

for customer_id in test_customers:

    print()
    print("-" * 50)
    print(
        "Customer:",
        customer_id
    )
    print("-" * 50)

    recommendations = (
        recommend_for_customer(
            customer_id,
            limit=5
        )
    )

    for item in recommendations:
        print(item)
```

```python
recommendation_rows = []


for customer_id in purchase_matrix.index:

    recommendations = (
        recommend_for_customer(
            customer_id,
            limit=5
        )
    )

    for item in recommendations:

        recommendation_rows.append({

            "CustomerID": customer_id,

            "ProductID": item["ProductID"],

            "Score": item["Score"]

        })


recommendation_results = pd.DataFrame(
    recommendation_rows
)

recommendation_results.head(20)
```

```python
recommendation_results.to_excel(
    "../datasets/personalized_recommendations.xlsx",
    index=False
)

print(
    "Personalized recommendations saved."
)
```
