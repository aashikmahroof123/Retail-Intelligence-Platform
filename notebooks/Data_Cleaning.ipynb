```python
import pandas as pd
```

```python
file_path = r"C:\Users\RSCH\Desktop\DWD\Final project\Retail-Intelligence-Platform\datasets\amez_dataset.xlsx"

data = pd.read_excel(
    file_path,
    sheet_name="Overall_Transactions"
)

data.head()
```

```python
print("Rows:", len(data))
print("Columns:", len(data.columns))
```

```python
print(data.columns.tolist())
```

```python
missing_values = data.isnull().sum()

print(missing_values)
```

```python
duplicate_count = data.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
duplicates = data[data.duplicated(keep=False)]

duplicates
```

```python
duplicates[
    [
        "InvoiceID",
        "InvoiceDate",
        "CustomerID",
        "ProductID",
        "ProductName",
        "Quantity",
        "UnitPrice",
        "Discount",
        "TotalAmount"
    ]
]
```

```python
data = data.drop_duplicates()
```

```python
print("Rows after removing duplicates:", len(data))
```

```python
print("Duplicates remaining:", data.duplicated().sum())
```

```python
print(data.isnull().sum().sum())
```

```python
print(data.isnull().sum().sum())
```

```python
data["InvoiceDate"] = pd.to_datetime(data["InvoiceDate"])

print("First date:", data["InvoiceDate"].min())
print("Last date:", data["InvoiceDate"].max())
```

```python
customer_invoice_count = (
    data.groupby("CustomerID")["InvoiceID"]
    .nunique()
)

print(customer_invoice_count.describe())
```

```python
products_per_invoice = (
    data.groupby("InvoiceID")["ProductID"]
    .nunique()
)

print(products_per_invoice.describe())
```

```python
multiple_product_invoices = (
    products_per_invoice > 1
).sum()

print(
    "Invoices containing multiple products:",
    multiple_product_invoices
)
```

```python
print(data[["Quantity", "UnitPrice", "Discount", "TotalAmount", "StockQuantity"]].describe())
```

```python
data = data[data["Quantity"] > 0]
```

```python
data = data[data["UnitPrice"] >= 0]
data = data[data["Discount"] >= 0]
data = data[data["StockQuantity"] >= 0]
```

```python
calculated_total = (
    data["UnitPrice"] * data["Quantity"]
    - data["Discount"]
)

difference = (data["TotalAmount"] - calculated_total).abs()

print("Incorrect TotalAmount:", (difference > 0.01).sum())
```

```python
data = data.dropna(
    subset=["InvoiceID", "InvoiceDate","CustomerID", "ProductID","ProductName","Quantity", "UnitPrice", "Discount", "TotalAmount", "StockQuantity"]
)
```

```python
data = data.dropna(
    subset=["Category", "SubCategory",]
)
```

```python
print("Rows:", len(data))
print("Columns:", len(data.columns))
print("Missing values:", data.isnull().sum().sum())
print("Duplicate rows:", data.duplicated().sum())
print("Invoices:", data["InvoiceID"].nunique())
print("Customers:", data["CustomerID"].nunique())
print("Products:", data["ProductID"].nunique())
```

```python
data.isnull().sum()
```

```python
data = data.dropna(
    subset=["StoreID", "StoreName",]
)
```

```python
data.isnull().sum()
```

```python
print("Rows:", len(data))
print("Columns:", len(data.columns))
print("Missing values:", data.isnull().sum().sum())
print("Duplicate rows:", data.duplicated().sum())
print("Invoices:", data["InvoiceID"].nunique())
print("Customers:", data["CustomerID"].nunique())
print("Products:", data["ProductID"].nunique())
```

```python
total_invoices = data["InvoiceID"].nunique()

print("Total number of invoices:", total_invoices)
```

```python
total_customers = data["CustomerID"].nunique()



print("Total number of customers:", total_customers)
```

```python
cleaned_file = "C:/Users/RSCH/Desktop/DWD/Final project/Retail-Intelligence-Platform/datasets/amez_dataset_cleaned.xlsx"

data.to_excel(
    cleaned_file,
    sheet_name="Overall_Transactions",
    index=False
)

print("Cleaned dataset saved successfully.")
```
